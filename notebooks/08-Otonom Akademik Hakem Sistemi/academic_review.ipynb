```python
import os
import yaml
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
# GÜNCELLEME: langchain_core kullanıyoruz
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

# --- ADIM 2: HAKEMLİK MOTORU ---
def load_prompt(name):
    with open("review_prompts.yaml", "r") as f:
        data = yaml.safe_load(f)
    return ChatPromptTemplate.from_template(data[name]["template"])

def run_review_system(text):
    if not os.getenv("GOOGLE_API_KEY"):
        print("HATA: API Anahtarı eksik.")
        return None, None, None

    model = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0.2)
    
    # 3 Farklı Ajanı Sırayla Çalıştır
    print("--- Akademik Hakem Sistemi Çalışıyor ---")
    
    print("1. Teknik Analiz...")
    # LCEL Yapısı: Prompt | Model | Parser
    tech_chain = load_prompt("technical_reviewer") | model | StrOutputParser()
    tech_rep = tech_chain.invoke({"manuscript": text})
    
    print("2. Metodoloji Analizi...")
    meth_chain = load_prompt("methodology_reviewer") | model | StrOutputParser()
    meth_rep = meth_chain.invoke({"manuscript": text})
    
    print("3. Dil Düzenlemesi...")
    edit_chain = load_prompt("copy_editor") | model | StrOutputParser()
    edit_rep = edit_chain.invoke({"manuscript": text})
    
    return tech_rep, meth_rep, edit_rep

# --- ADIM 3: TEST ---
kusurlu_makale = """
We propose DeepFast. It completely solves the traveling salesman problem. 
It is significantly better than all methods. 
We trained it on some data and it runs super fast on my laptop.
"""

if __name__ == "__main__":
    t, m, e = run_review_system(kusurlu_makale)
    
    if t:
        print("\n=== Rapor 1: Teknik ===\n", t)
        print("\n=== Rapor 2: Metodoloji ===\n", m)
        print("\n=== Rapor 3: Editör ===\n", e)
```
