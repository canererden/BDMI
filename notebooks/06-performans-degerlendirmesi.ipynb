```python
import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

reference = [["Harika", "bir", "gün"]]
candidate = ["Mükemmel", "bir", "hava"]

smooth = SmoothingFunction().method1
bleu_score = sentence_bleu(reference, candidate, smoothing_function=smooth)

print("BLEU Skoru:", bleu_score)
```

```python
from rouge_score import rouge_scorer

reference = "Harika bir gün"
candidate = "Mükemmel bir hava"

scorer = rouge_scorer.RougeScorer(['rouge1', 'rougeL'], use_stemmer=True)
scores = scorer.score(reference, candidate)

print("ROUGE Skorları:", scores)
```

```python
import math

probabilities = [0.25, 0.5, 0.125, 0.125]  # örnek olasılıklar

N = len(probabilities)
perplexity = math.pow(math.prod([1/p for p in probabilities]), 1/N)

print("Perplexity:", perplexity)
```

```python
from bert_score import score

cands = ["Mükemmel bir hava"]
refs = ["Harika bir gün"]

P, R, F1 = score(cands, refs, lang="tr", verbose=True)

print("BERTScore Precision:", P.mean().item())
print("BERTScore Recall:", R.mean().item())
print("BERTScore F1:", F1.mean().item())
```

```python
def g_eval_simulation(text):
    reasoning = f"Metin: '{text}'\n- Akıcılık: iyi\n- Tutarlılık: orta\n- Alaka düzeyi: yüksek"
    score = 4.2  # örnek puan
    return reasoning, score

reasoning, score = g_eval_simulation("Bugün hava çok güzel, dışarı çıkmak için harika bir zaman.")
print("G-Eval Düşünce Zinciri:\n", reasoning)
print("G-Eval Skoru:", score)
```
